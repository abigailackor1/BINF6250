```python
class HMM:
    def __init__(self, init_probs, trans_probs, emission_probs):
        self.init_probs = init_probs
        self.trans_probs = trans_probs
        self.emission_probs = emission_probs

    def Viterbi(self, obs):
        best_prob = [{}]   # best_prob[t][state] = best path probability ending in state at t
        back = [{}]        # back[t][state] = previous state on that best path

        # STEP 1: initialize (position 0)
        for state in self.init_probs:
            start_chance = self.init_probs[state]
            emit_chance = self.emission_probs[state][obs[0]]
            best_prob[0][state] = start_chance * emit_chance
            back[0][state] = None

        # STEP 2: recursion (positions 1 to the end)
        for t in range(1, len(obs)):
            best_prob.append({})
            back.append({})

            for state in self.init_probs:
                best_score = 0
                best_prev = None

                for prev_state in self.init_probs:
                    score = best_prob[t - 1][prev_state] * self.trans_probs[prev_state][state]
                    if score > best_score:
                        best_score = score
                        best_prev = prev_state

                
                best_prob[t][state] = best_score * self.emission_probs[state][obs[t]]
                back[t][state] = best_prev

        # STEP 3: terminate
        last = len(obs) - 1
        final_scores = best_prob[last]
        best_final_state = max(final_scores, key=final_scores.get)
        best_path_prob = final_scores[best_final_state]

        # STEP 4: traceback
        path = [best_final_state]
        for t in range(last, 0, -1):
            path.append(back[t][path[-1]])
        path.reverse()          

        return path, best_path_prob   
```

```python
# Example observation sequence
obs = "GGCACTGAA"

# Example initial probabilities (probability of starting in each state)
init_probs = {
    "I": 0.2,
    "G": 0.8
}

# Example transition probabilities (probability of moving from one state to another)
trans_probs = {
    "I": {"I": 0.7, "G": 0.3},
    "G": {"I": 0.1, "G": 0.9}
}

# Example emission probabilities (probability of observing a symbol in a given state)
emit_probs = {
    "I": {"A": 0.1, "C": 0.4, "G": 0.4, "T": 0.1},
    "G": {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3}
}
```

```python
hmm = HMM(init_probs, trans_probs, emit_probs)
```

```python
path,prob = hmm.Viterbi(obs)
print("".join(path), prob)
```
